```python
taxafile="full_taxa_draft.csv"
taxonomy_draft=pd.read_csv(taxafile, usecols=lambda x: x != 'Unnamed: 0')
taxonomy_draft.head()
```

```python
# help from gemini
# Function to fetch standard lineage for a given TaxID
def fill_missing_taxonomy(row):
    tax_id = row['otuid']  # Using the 'otu' column as TaxID
    
    # Ranks matching your DataFrame columns
    target_ranks = ['domain','kingdom', 'phylum', 'class', 'order', 'family', 'genus', 'species']
    
    try:
        taxon = taxopy.Taxon(int(tax_id), taxdb)
        # NCBI calls Domain 'superkingdom'
        rank_dict = taxon.rank_name_dictionary
        if 'superkingdom' in rank_dict:
            rank_dict['domain'] = rank_dict.pop('superkingdom')
            
        # Update missing/NaN values in the row only
        for rank in target_ranks:
            if pd.isna(row.get(rank)) or row.get(rank) == '':
                row[rank] = rank_dict.get(rank, 'Unclassified')
                
    except Exception:
        # If TaxID is obsolete or unassigned
        for rank in target_ranks:
            if pd.isna(row.get(rank)):
                row[rank] = 'Unclassified'
                
    return row

# convert otuid to integer
taxonomy_draft['otuid'] = taxonomy_draft['otuid'].astype(str).str.extract(r'(\d+)')[0]

# Update the DataFrame row by row
taxonomy_filled = taxonomy_draft.apply(fill_missing_taxonomy, axis=1)

# Inspect results
taxonomy_filled.head()
```

```python
# example spp to check: 3398215, 3398210, 1435008
check=[3398215, 3398210, 1435008, 3059980, 134287]
check_otus_str = [str(i) for i in check]
taxonomy_filled[taxonomy_filled['otuid'].astype(str).isin(check_otus_str)]
```

```python
# If kingdom ends with 'virae', set domain to 'Virus'
# or if species name contains virus - also change kingdom to virus
virus_mask = taxonomy_filled['kingdom'].astype(str).str.endswith('virae', na=False)
taxonomy_filled.loc[virus_mask, 'domain'] = 'Virus'

virus_spp = taxonomy_filled['species'].astype(str).str.contains('virus', na=False)
taxonomy_filled.loc[virus_spp, 'domain'] = 'Virus'
```
